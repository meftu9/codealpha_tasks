```python
import os
import pandas as pd
import numpy as np
import librosa
import matplotlib.pyplot as plt

DATASET_PATH = "../dataset/ravdess"

print(os.listdir(DATASET_PATH))
```

```python
emotion_labels = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

data = []

for actor in os.listdir(DATASET_PATH):
    actor_path = os.path.join(DATASET_PATH, actor)

    for filename in os.listdir(actor_path):
        if filename.endswith(".wav"):
            emotion_code = filename.split("-")[2]

            emotion = emotion_labels[emotion_code]

            file_path = os.path.join(actor_path, filename)

            data.append({
                "file_path": file_path,
                "filename": filename,
                "emotion": emotion
            })

df = pd.DataFrame(data)

print(df.head())
print("Total audio files:", len(df))
```

```python
TARGET_SAMPLE_RATE = 22050

audio_samples = []
audio_durations = []

for file_path in df["file_path"]:
    audio, sr = librosa.load(
        file_path,
        sr=TARGET_SAMPLE_RATE
    )

    audio_samples.append(audio)
    audio_durations.append(len(audio) / sr)

print("Processed audio files:", len(audio_samples))
print("Target sample rate:", TARGET_SAMPLE_RATE)
print("Shortest duration:", min(audio_durations))
print("Longest duration:", max(audio_durations))
```

```python
plt.figure(figsize=(10, 4))

plt.hist(audio_durations, bins=30)

plt.title("Audio Duration Distribution")
plt.xlabel("Duration (seconds)")
plt.ylabel("Number of files")

plt.show()
```
