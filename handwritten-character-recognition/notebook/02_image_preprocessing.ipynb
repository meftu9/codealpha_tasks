# 02 - Image Preprocessing: Numbers & Letters

This notebook covers the data transformation pipeline for character and letter recognition:
1. Rescaling pixel values to [0.0, 1.0].
2. Reshaping dimensions to add the channel dimension `(N, 28, 28, 1)`.
3. Checking image orientation and color inversion (dark background vs light background).
4. Preparing train, validation, and test splits.
5. Caching preprocessed arrays to `models/character_features.npz`.


```python
# Import libraries
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps

sys.path.append(os.path.abspath(".."))
from src.preprocessing import (
    load_raw_data,
    preprocess_pipeline,
    save_preprocessed_data,
    preprocess_single_image
)

print("Preprocessing modules loaded successfully!")
```

## 1. Load Raw Dataset


```python
raw_data_path = os.path.abspath("../dataset/raw/emnist_balanced.npz")
(x_train_raw, y_train), (x_test_raw, y_test) = load_raw_data(raw_data_path, dataset_type="emnist")
```

## 2. Normalization & Reshaping


```python
x_train, x_test = preprocess_pipeline(x_train_raw, x_test_raw)

print(f"Preprocessed x_train shape: {x_train.shape}, dtype: {x_train.dtype}")
print(f"Preprocessed x_test shape:  {x_test.shape}, dtype: {x_test.dtype}")
print(f"Min pixel value: {x_train.min()} | Max pixel value: {x_train.max()}")
```

## 3. Orientation & Inversion Handling for Custom User Inputs


```python
# Test real-world inversion logic on a letter sample
letter_sample = x_train_raw[100]

# Simulate dark ink on white paper
paper_drawing = 255 - letter_sample

# Apply preprocessing
tensor, processed_img = preprocess_single_image(paper_drawing)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(letter_sample, cmap="gray")
axes[0].set_title("Standard Dataset (White on Black)")
axes[0].axis("off")

axes[1].imshow(paper_drawing, cmap="gray")
axes[1].set_title("Simulated User Drawing (Dark on White)")
axes[1].axis("off")

axes[2].imshow(tensor.squeeze(), cmap="gray")
axes[2].set_title("Auto-Inverted & Normalized")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```

## 4. Save Preprocessed Data


```python
features_output = os.path.abspath("../models/character_features.npz")
save_preprocessed_data(x_train, y_train, x_test, y_test, features_output)
```

## Summary
- Pixel values scaled to `[0.0, 1.0]`.
- Channels added: `(N, 28, 28, 1)`.
- Preprocessed tensors saved into `models/character_features.npz`.
