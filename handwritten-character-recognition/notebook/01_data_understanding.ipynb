# 01 - Data Understanding: Handwritten Number & Letter Recognition

This notebook explores the dataset for both handwritten digits (0–9) and letters (A–Z / a–z) using **EMNIST Balanced** and **MNIST**.
Key stages covered:
1. Loading the dataset (112,800 training and 18,800 testing character images).
2. Checking dataset sizes, dimensions, and data types across 47 classes.
3. Visualizing character distributions across digits and letters.
4. Previewing image samples with grayscale plots.
5. Analyzing pixel intensity distributions.


```python
# Import essential libraries
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

sys.path.append(os.path.abspath(".."))
from src.preprocessing import load_raw_data, load_label_mapping

print("TensorFlow Version:", tf.__version__)
print("Libraries imported successfully!")
```

## 1. Load EMNIST Balanced Dataset (Letters + Numbers)
We load the 47-class dataset using `load_raw_data(dataset_type='emnist')`.


```python
raw_data_path = os.path.abspath("../dataset/raw/emnist_balanced.npz")
(x_train, y_train), (x_test, y_test) = load_raw_data(raw_data_path, dataset_type="emnist")
label_map = load_label_mapping(os.path.abspath("../models/label_mapping.json"))

print(f"Training images: {x_train.shape}, Labels: {y_train.shape}")
print(f"Testing images:  {x_test.shape}, Labels: {y_test.shape}")
print(f"Total Character Classes: {len(np.unique(y_train))}")
print(f"Image resolution: {x_train.shape[1]}x{x_train.shape[2]} pixels")
```

## 2. Character & Class Distribution Analysis
Checking balance across all 47 classes (Digits 0–9, Uppercase A–Z, and distinctive lowercases).


```python
unique_classes, counts_train = np.unique(y_train, return_counts=True)
class_labels = [label_map.get(i, str(i)) for i in unique_classes]

plt.figure(figsize=(16, 5))
sns.barplot(x=class_labels, y=counts_train, palette="viridis")
plt.title("Character Distribution across 47 Classes (EMNIST Balanced)", fontsize=14, fontweight="bold")
plt.xlabel("Character Class (0-9, A-Z, a-z)", fontsize=12)
plt.ylabel("Number of Training Samples", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print(f"Each character class has exactly {counts_train[0]} training samples (perfectly balanced).")
```

## 3. Visualizing Sample Handwritten Numbers and Letters
Previewing a representative 4×5 grid of handwritten numbers and letters.


```python
fig, axes = plt.subplots(4, 5, figsize=(12, 10))
rng = np.random.default_rng(42)
indices = rng.choice(len(x_train), size=20, replace=False)

for ax, idx in zip(axes.flat, indices):
    char_label = label_map.get(y_train[idx], str(y_train[idx]))
    ax.imshow(x_train[idx], cmap="gray")
    ax.set_title(f"Class {y_train[idx]}: '{char_label}'", fontsize=12, fontweight="bold")
    ax.axis("off")

plt.suptitle("Sample Handwritten Characters: Numbers & Letters", fontsize=15, fontweight="bold", y=0.95)
plt.tight_layout()
plt.show()
```

## 4. Pixel Intensity Range & Statistics


```python
print(f"Minimum pixel value: {x_train.min()}")
print(f"Maximum pixel value: {x_train.max()}")
print(f"Mean pixel intensity: {x_train.mean():.2f}")
print(f"Standard deviation:  {x_train.std():.2f}")
```
