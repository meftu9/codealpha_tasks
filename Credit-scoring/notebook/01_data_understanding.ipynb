# 01 - Data Understanding: Credit Scoring Model

This notebook explores the raw financial history dataset (`dataset/raw/train.csv`) used to predict individual creditworthiness (**Poor**, **Standard**, **Good**).

Key stages covered:
1. Loading the raw 100,000-record dataset via `src.data_loader`.
2. Inspecting schema, column types, and unique customer counts (`Customer_ID`).
3. Visualizing the `Credit_Score` target class distribution.
4. Auditing missing values, trailing underscores, and sentinel/corrupted entries.


```python
import os
import sys

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sys.path.append(os.path.abspath(".."))
from src.data_loader import load_raw_data, TARGET_COLUMN

sns.set_theme(style="whitegrid")
```

## 1. Load Raw Financial Dataset


```python
raw_path = os.path.abspath("../dataset/raw/train.csv")
df = load_raw_data(raw_path)
print(f"Unique Customers: {df['Customer_ID'].nunique()}")
df.head()
```

## 2. Target Class Distribution


```python
order = ["Poor", "Standard", "Good"]
counts = df[TARGET_COLUMN].value_counts().reindex(order)
print(counts)

fig, ax = plt.subplots(figsize=(7, 4))
sns.barplot(x=counts.index, y=counts.values, hue=counts.index, palette=["#e53e3e", "#dd6b20", "#2f855a"], legend=False, ax=ax)
ax.set_title("Credit Score Target Distribution", fontsize=13, fontweight="bold")
ax.set_ylabel("Records")
ax.set_xlabel("Credit Score")
plt.tight_layout()
plt.show()
```

## 3. Data Quality & Missing Value Audit


```python
audit = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "null_count": df.isna().sum(),
    "null_pct": (df.isna().mean() * 100).round(2),
    "nunique": df.nunique(),
})
audit
```
