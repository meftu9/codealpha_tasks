# 04 - Model Evaluation & Diagnostics

This notebook inspects the evaluation outputs generated by `python -m src.evaluate`, including Precision, Recall, F1-Score, One-vs-Rest ROC-AUC, confusion matrix, and top feature importances.


```python
import json
import os
import pandas as pd
from IPython.display import Image, display

summary_path = os.path.abspath("../reports/evaluation_summary.json")
with open(summary_path, "r", encoding="utf-8") as f:
    summary = json.load(f)

print("Best Model:", summary["best_model"])
rows = [
    {
        "Model": name,
        "Accuracy": round(m["accuracy"], 4),
        "Precision": round(m["weighted_precision"], 4),
        "Recall": round(m["weighted_recall"], 4),
        "F1-Score": round(m["weighted_f1_score"], 4),
        "ROC-AUC (OvR)": round(m["roc_auc_ovr_weighted"], 4),
    }
    for name, m in summary["models"].items()
]
pd.DataFrame(rows)
```

```python
for fname in ["model_comparison.png", "confusion_matrix.png", "roc_curves.png", "feature_importance.png"]:
    path = os.path.abspath(os.path.join("../reports", fname))
    if os.path.exists(path):
        print(fname)
        display(Image(filename=path))
```
